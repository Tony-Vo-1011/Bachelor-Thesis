```python
import pandas as pd
from pathlib import Path

# change this to your folder
base = Path("/Users/tonyvo/Desktop/Thesis/netlogo/data")

files = {
    "nodes": base / "roads" / "roads_nodes.csv",
    "edges": base / "roads" / "roads_from_to.csv",
    "places": base / "places" / "places.csv",
}
```

```python
for name, path in files.items():
    print("\n" + "="*80)
    print(f"{name.upper()} -> {path}")
    print("="*80)

    df = pd.read_csv(path)

    print("\nShape:")
    print(df.shape)

    print("\nColumns:")
    print(df.columns.tolist())

    print("\nDtypes:")
    print(df.dtypes)

    print("\nMissing values:")
    print(df.isna().sum())

    print("\nFirst 5 rows:")
    display(df.head())

    print("\nSample values per column:")
    for col in df.columns:
        vals = df[col].dropna().astype(str).head(3).tolist()
        print(f"{col}: {vals}")
```

```python
nodes_path = base / "roads" / "roads_nodes.csv"
nodes = pd.read_csv(nodes_path)

nodes_clean = nodes[["node_id", "x", "y"]].copy()

nodes_clean.to_csv(base / "roads" / "roads_nodes_clean.csv", index=False)
```

```python
edges_path = base / "roads" / "roads_from_to.csv"
edges = pd.read_csv(edges_path)

edges_clean = edges[["edge_id", "from_id", "to_id", "Shape_Length"]].copy()

print(edges_clean.head())
print(edges_clean.dtypes)
print(edges_clean.isna().sum())
print("Duplicate edge_id:", edges_clean["edge_id"].duplicated().sum())

edges_clean.to_csv(base / "roads" / "edges_clean.csv", index=False)
```

```python
places_path = base / "places" / "places.csv"
places = pd.read_csv(places_path)

places_clean = places[["id", "x", "y"]].copy()

print(places_clean.head())
print(places_clean.dtypes)
print(places_clean.isna().sum())

places_clean.to_csv(base / "places" / "places_clean.csv", index=False)
```
