```python
import pandas as pd
import numpy as np

input_path = r"C:/Users/15177459/Desktop/gis_outputs/cost_od_matrix/transport/input/ODMatrix_all_transport.csv"
output_path = r"C:/Users/15177459/Desktop/gis_outputs/cost_od_matrix/transport/output/ODMatrix_all_transport_zscore.csv"

df = pd.read_csv(input_path)

print("Columns in file:")
print(df.columns.tolist())
print("Order of transport modes: Cycle Drive Walk")
```

```python
cycle_col = "mean"
drive_col = "mean_1"
walk_col = "mean_12"

for col in [cycle_col, drive_col, walk_col]:
    df[col] = pd.to_numeric(df[col], errors ="coerce")
    df[col] = df[col].replace(0, np.nan)

print("\nSummary stats after cleaning:")
print(df[[cycle_col, drive_col, walk_col]].describe())
```

```python
df["cycle_z"] = (df[cycle_col]-df[cycle_col].mean()) / df[cycle_col].std()
df["drive_z"] = (df[drive_col]-df[drive_col].mean()) / df[drive_col].std()
df["walk_z"] = (df[walk_col]-df[walk_col].mean()) / df[walk_col].std()

df["cycle_access"] = -df["cycle_z"]
df["drive_access"] = -df["drive_z"]
df["walk_access"] = -df["walk_z"]


df["transport_access_score"] = df[["cycle_access", "drive_access", "walk_access"]].mean(axis=1)

print("\nPreview")
print(
    df[
        [walk_col, cycle_col, drive_col,
         "walk_access", "cycle_access", "drive_access",
         "transport_access_score"]
    ].head()
)

df.to_csv(output_path)
```
