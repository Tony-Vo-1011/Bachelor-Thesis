# Transport Accessibility

This notebook aggregates the mean transport time to all other neighbourhoods, per transport mode. Transport accessibility was defined as how connected a neighbourhood was to all other neighbourhoods.





```python
import pandas as pd

base = "C:/Users/15177459/Desktop/gis_outputs/cost_od_matrix/transport/input/"
```

```python
od_matrix = ['drive', 'cycle', 'walk']
for i in range(len(od_matrix)):
    od_matrix[i] = f"ODMatrix_buurtbuurt_{od_matrix[i]}.csv"

print(od_matrix)
```

```python

a = pd.read_csv(base + od_matrix[0])
print(a.columns)
```

```python
def third_places_per_origin(mode, cost):
    od =  pd.read_csv(base + mode)
    origin_col = "OriginID"

    summary_count = (
        od.groupby(origin_col)
        .agg(
            mean=(cost, "mean"),
            median=(cost, "median"),
            std = (cost, "std"),
            nearest = (cost, "min")
        )
        .reset_index()
    )
    return summary_count
```

```python
cost_list = [ 'Total_DrvingTime',  'Total_BikeTime', 'Total_WalkTime']

for i in range(len(od_matrix)):
    summary = third_places_per_origin(od_matrix[i], cost_list[i])
    summary.to_csv("C:/Users/15177459/Desktop/gis_outputs/cost_od_matrix/transport/output/" + od_matrix[i])
    print
```
