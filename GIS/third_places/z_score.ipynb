```python
import pandas as pd
import numpy as np

walk_path = r"C:/Users/15177459/Desktop/gis_outputs/cost_od_matrix/third_places/output/ODMatrix_walking_10.csv"
cycle_path = r"C:/Users/15177459/Desktop/gis_outputs/cost_od_matrix/third_places/output/ODMatrix_cycling_10.csv"
drive_path = r"C:/Users/15177459/Desktop/gis_outputs/cost_od_matrix/third_places/output/ODMatrix_driving_10.csv"


walk = pd.read_csv(walk_path)
cycle = pd.read_csv(cycle_path)
drive = pd.read_csv(drive_path)

print("Walk columns:", walk.columns.tolist())
print("Cycle columns:", cycle.columns.tolist())
print("Drive columns:", drive.columns.tolist())
```

```python
id_col = "OriginID"
count_col = "count"

walk = walk[[id_col, count_col]].rename(columns={count_col: "walk_10_count"})
cycle = cycle[[id_col, count_col]].rename(columns={count_col: "cycle_10_count"})
drive = drive[[id_col, count_col]].rename(columns={count_col: "drive_10_count"})

df = walk.merge(cycle, on=id_col, how="outer")
df = df.merge(drive, on=id_col, how="outer")

df[["walk_10_count", "cycle_10_count", "drive_10_count"]] = (
    df[["walk_10_count", "cycle_10_count", "drive_10_count"]].fillna(0)
)

print(df.head())
```

```python
for col in ["walk_10_count", "cycle_10_count", "drive_10_count"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
df["walk_tp_z"] = (df["walk_10_count"]-df["walk_10_count"].mean()) / df["walk_10_count"].std()
df["cycle_tp_z"] = (df["cycle_10_count"]-df["cycle_10_count"].mean()) / df["cycle_10_count"].std()
df["drive__tp_z"] = (df["drive_10_count"]-df["drive_10_count"].mean()) / df["drive_10_count"].std()

df["third_place_access_score"] = df[["cycle_tp_z", "drive__tp_z", "walk_tp_z"]].mean(axis=1)

print("\nPreview")
print(
    df[
        [id_col, "walk_10_count", "cycle_10_count", "drive_10_count", "third_place_access_score"]
    ].head()
)

df.to_csv("C:/Users/15177459/Desktop/gis_outputs/cost_od_matrix/third_places/output/third_place_z_score.csv")
```
