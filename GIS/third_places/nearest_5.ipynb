```python
costs = ['Total_DrvingTime','Total_BikeTime', 'Total_WalkTime']

def average_time_5(od, cost):
    origin_col = "OriginID"
    cost_col = cost

    summary_nearest = (
        od.groupby(origin_col)
        .agg(
            avg_time = (cost_col, "mean")
        )
        .reset_index()
    )
    return summary_nearest

for i in range(len(paths)):
    a = pd.read_csv(base + paths[i])
    summary = average_time_5(a, costs[i])
    summary.to_csv("C:/Users/15177459/Desktop/gis_outputs/cost_od_matrix/output/ODmatrix" + paths[i])
```
