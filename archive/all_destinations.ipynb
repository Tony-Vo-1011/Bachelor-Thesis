```python
def third_places_per_origin(od, cost):
    origin_col = "OriginID"
    cost_col = cost

    summary_count = (
        od.groupby(origin_col)
        .agg(
            count=(origin_col, "size"),
            nearest_min=(cost_col, "min")
        )
        .reset_index()
    )
    return summary_count
```

```python
cost_list = [ 'Total_DrvingTime', 'Total_DrvingTime',  'Total_BikeTime', 'Total_BikeTime', 'Total_WalkTime','Total_WalkTime']
i = 0

for keys, values in od_matrix.items():
    for j in range(len(values)):
        a = pd.read_csv(base + od_matrix[keys][j])
        summary = third_places_per_origin(a, cost_list[i])
        summary.to_csv("C:/Users/15177459/Desktop/gis_outputs/cost_od_matrix/output/" + od_matrix[keys][j])
        i += 1


```
